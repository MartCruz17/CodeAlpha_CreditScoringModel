```python
late_cols = ["NumberOfTime30-59DaysPastDueNotWorse",
             "NumberOfTimes90DaysLate",
             "NumberOfTime60-89DaysPastDueNotWorse"]

# Flag the rows before touching them
df["HadSuspiciousLateCode"] = (df[late_cols] >= 96).any(axis=1).astype(int)

# Cap all three columns at a believable max (17 was the highest real value we saw)
for col in late_cols:
    df[col] = df[col].clip(upper=17)
```

```python
df = df.drop(columns=["Unnamed: 0"])

X = df.drop(columns=["SeriousDlqin2yrs"])
y = df["SeriousDlqin2yrs"]
```

```python
print(X.shape)
print(y.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
import pandas as pd
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)
print(X_scaled.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

print(X_train.shape, X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, model.predict_proba(X_test)[:, 1]))
```

```python
model_balanced = LogisticRegression(class_weight="balanced")
model_balanced.fit(X_train, y_train)

y_pred_balanced = model_balanced.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_balanced))
print("Precision:", precision_score(y_test, y_pred_balanced))
print("Recall:", recall_score(y_test, y_pred_balanced))
print("F1 Score:", f1_score(y_test, y_pred_balanced))
print("ROC-AUC:", roc_auc_score(y_test, model_balanced.predict_proba(X_test)[:, 1]))
```

```python
from sklearn.ensemble import RandomForestClassifier

model_rf = RandomForestClassifier(class_weight="balanced", random_state=42)
model_rf.fit(X_train, y_train)

y_pred_rf = model_rf.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print("Precision:", precision_score(y_test, y_pred_rf))
print("Recall:", recall_score(y_test, y_pred_rf))
print("F1 Score:", f1_score(y_test, y_pred_rf))
print("ROC-AUC:", roc_auc_score(y_test, model_rf.predict_proba(X_test)[:, 1]))
```

```python
model_rf2 = RandomForestClassifier(class_weight="balanced_subsample", random_state=42)
model_rf2.fit(X_train, y_train)

y_pred_rf2 = model_rf2.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_rf2))
print("Precision:", precision_score(y_test, y_pred_rf2))
print("Recall:", recall_score(y_test, y_pred_rf2))
print("F1 Score:", f1_score(y_test, y_pred_rf2))
print("ROC-AUC:", roc_auc_score(y_test, model_rf2.predict_proba(X_test)[:, 1]))
```
